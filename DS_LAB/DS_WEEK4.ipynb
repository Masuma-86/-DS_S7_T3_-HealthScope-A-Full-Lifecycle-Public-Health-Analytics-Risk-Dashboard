```python
import matplotlib.pyplot as plt
y1 = [] #stores #ve values
y2 = [] #stores -ve vakes
x = range(-100,100,10)
for i in x: y1.append(i**2) 
for i in x: y2.append(-i**2)

plt.plot(x, y1)
plt.plot(x, y2)
plt.xlabel("x")
plt.xlabel("y")
plt.ylim(-2000, 2000)
plt.axhline(0)
plt.axhline(0)
plt.savefig("quad.png")
plt.show()
            
```
